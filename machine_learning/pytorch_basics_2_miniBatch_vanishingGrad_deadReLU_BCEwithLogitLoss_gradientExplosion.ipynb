```python
# 6: Training on a "Mini-Batch"The Concept: In the real world, we don't train on one row of data at a time. 
# we use a "batch."The Task: 
# 1. Create a dataset X with 3 rows of data: [[1.0, 2.0], [2.0, 3.0], [3.0, 4.0]].
# 2. Create $Y_{true}$ with 3 targets: [[5.0], [8.0], [11.0]].
# 3. Train your nn.Linear(2, 1) model for 50 epochs.
# Practical Goal: Observe how nn.MSELoss automatically calculates the average error across all 3 rows.

#import necessary libraries 
import torch
from torch.nn import Linear, MSELoss
from torch.utils.data import TensorDataset, DataLoader
from torch.optim import SGD


X = torch.tensor([[1.0, 2.0], [2.0, 3.0], [3.0, 4.0]]) #required_grad = True is not required here as model itself initializer gradients
y_true = torch.tensor([[5.0], [8.0], [11.0]])

#dataset and loader are formed to perform minibatch
dataset = TensorDataset(X,y_true)
loader = DataLoader(dataset, batch_size = 2)
print('input tensor X is: ',X)

#setup model and required loss functions
model = Linear(in_features = 2, out_features=1)
loss = MSELoss()
optimizer = SGD(model.parameters(), lr = 0.01)

print('\n')
print('\n')


#initialize standard training loop with learning rate 0.01
for epoch in range(1,51):
    print('epoch iteration number : ', epoch)
    print('model weights: ', model.weight)
    
    for batch_X,batch_y in loader:
        #use model to predict
        y_pred = model(batch_X)

        #calculate the loss made in prediction
        mean_squared_error = loss(y_pred, batch_y)
    
        #reset the gradients so that they dont add up from previous epoch
        optimizer.zero_grad()
        
        #backpropagation
        mean_squared_error.backward()

        #adjust model weights to that model can predict better the next time.
        optimizer.step()
        print('y_pred is: ', y_pred)
    print(f'mean_squared_error for epoch number {epoch} is :', mean_squared_error)
    print('optimzed weights:', model.weight)
    print('\n')


```

```python
# 7: The "Vanishing Gradient" Sabotage
# The Concept: If your Learning Rate is too small, or your inputs are massive, gradients can become tiny.
# The Task: 
# 1. Use the same code from Question 5, but change your Learning Rate to something extremely small: lr = 1e-10.
# 2. Run it for 20 epochs.
# 3. Print the model.weight.grad in each loop.

# Practical Goal: See what happens when the gradient exists, but the "step" is so small the weights barely move. How many epochs would it take to actually learn at this speed?



#import necessary libraries 
import torch
from torch.nn import Linear, MSELoss
from torch.utils.data import TensorDataset, DataLoader
from torch.optim import SGD


X = torch.tensor([[1.0, 2.0], [2.0, 3.0], [3.0, 4.0]]) #required_grad = True is not required here as model itself initializer gradients
y_true = torch.tensor([[5.0], [8.0], [11.0]])

#dataset and loader are formed to perform minibatch
dataset = TensorDataset(X,y_true)
loader = DataLoader(dataset, batch_size = 2)
print('input tensor X is: ',X)

#setup model and required loss functions
model = Linear(in_features = 2, out_features=1)
loss = MSELoss()
optimizer = SGD(model.parameters(), lr = 1e-10)

print('\n')
print('\n')


#initialize standard training loop with learning rate 0.01
for epoch in range(1,21):
    print('epoch iteration number : ', epoch)
    print('model weights: ', model.weight)
    
    for batch_X,batch_y in loader:
        #use model to predict
        y_pred = model(batch_X)

        #calculate the loss made in prediction
        mean_squared_error = loss(y_pred, batch_y)
    
        #reset the gradients so that they dont add up from previous epoch
        optimizer.zero_grad()
        
        #backpropagation
        mean_squared_error.backward()

        #adjust model weights to that model can predict better the next time.
        optimizer.step()
        print('y_pred is: ', y_pred)
    print(f'mean_squared_error for epoch number {epoch} is: ', mean_squared_error)
    print('optimzed weights: ', model.weight)
    print('model weight gradient: ', model.weight.grad)
    print('\n')
```

```python
# 9: Binary Classification (Cross Entropy)The Concept: Not everything is a regression (predicting a price). Sometimes we predict "Yes" (1) or "No" (0). For this, we use nn.BCEWithLogitsLoss.
# The Task: 
# 1. Change your $Y_{true}$ to torch.tensor([[1.0]]).
# 2. Change your loss function to criterion = nn.BCEWithLogitsLoss().
# 3. Run the training loop for 20 epochs.
# Practical Goal: Notice how the loss behaves differently compared to MSE.


import torch
from torch.nn import BCEWithLogitsLoss, Linear
from torch.optim import SGD
from colorama import Fore,Style

X = torch.tensor([-10.0, -5.0])
y_true = torch.tensor([1.0])

model = Linear(in_features = 2, out_features=1)
loss = BCEWithLogitsLoss()
optimizer = SGD(model.parameters(), lr = 0.01)


for epoch in range(0,20):
    print(f'iteration for epoch number: ', epoch)
    optimizer.zero_grad() 

    y_pred = model(X)
    error = loss(y_pred, y_true)
    print(Fore.GREEN + 'y_pred is: ', y_pred)
    print(Style.RESET_ALL)
    print('error is: ', error)
    print(Fore.GREEN + 'model weights: ', model.weight)
    print(Style.RESET_ALL)

    error.backward()
    optimizer.step()
    print('model weight gradient: ', model.weight.grad)

    print(Fore.GREEN + 'optimzed model weights: ', model.weight)
    print(Style.RESET_ALL)

    print('\n')

```
